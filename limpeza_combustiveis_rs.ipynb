```python
df_rs = df[df["Estado - Sigla"] == "RS"].copy()
df_rs.shape
```

```python
df_rs.columns = (df_rs.columns.str.strip().str.lower()
                 .str.replace(" - ", "_").str.replace(" ", "_"))
```

```python
df_rs["valor_de_venda"] = pd.to_numeric(df_rs["valor_de_venda"].str.replace(",", "."), errors="coerce")
#Transformação de string para tipo numérico
```

```python
df_rs = df_rs.drop(columns=["valor_de_compra"])
```

```python
df_rs["data_da_coleta"].head()
```

```python
df_rs["data_da_coleta"] = pd.to_datetime(df_rs["data_da_coleta"], dayfirst=True, errors="coerce")
```
